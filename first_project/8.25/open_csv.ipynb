```python
import pandas as pd
import glob
import os

folder_path = 'C:/Users/user/Desktop/데이터들/subway'

file_list = glob.glob(os.path.join(folder_path, '*.csv'))

df_list = []

for file in file_list:

    # 기존에 만든 통합 파일은 제외
    if 'subway_2호선_통합' in os.path.basename(file):
        continue

    df = None

    # 인코딩 자동 확인
    for encoding in ['utf-8-sig', 'utf-8', 'cp949', 'euc-kr']:
        try:
            df = pd.read_csv(
                file,
                encoding=encoding,
                index_col=False
            )
            print(f'{os.path.basename(file)} → {encoding}')
            break
        except UnicodeDecodeError:
            continue

    if df is None:
        print(f'읽기 실패: {file}')
        continue

    # 노선명 컬럼이 없는 파일 제외
    if '노선명' not in df.columns:
        print(f'노선명 없음 → {os.path.basename(file)}')
        continue

    # 2호선만 추출
    df = df[df['노선명'].astype(str).str.strip() == '2호선'].copy()

    if df.empty:
        continue

    # 사용일자에서 연도 추출
    df['연도'] = df['사용일자'].astype(str).str[:4]

    # 필요한 컬럼만 남기기
    df = df[
        [
            '연도',
            '노선명',
            '역명',
            '승차총승객수',
            '하차총승객수'
        ]
    ]

    df_list.append(df)


# 모든 파일 통합
result = pd.concat(df_list, ignore_index=True)

# 확인
print(result.head())
print(result.shape)
print(result.columns)
```
